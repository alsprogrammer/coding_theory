## Сжатие информации

В этой работе мы собираемся реализовать алгоритмы сжатия информации.

> Замечание: Следует различать **преобразование**, **сжатие без потерь**, и **сжатие с потерями**. 

Преобразование не уменьшает объем информации - оно лишь меняет ее представление. Например, значения амплитуд могут быть *представлены* собственно амплитудами, а могут быть *представлены* как разность с предыдущим значением амплитуды. Это - представление.

Сжатие пакует информацию плотнее, позволяя уместить ту же инфоормацию в меньшее количество байт. Гранб тут довольно тонкая, и в большей степени это вопрос терминологии.

И тот, и другой подход позволяют восстановить исходное сообщение один к одному.

Сжатие с потерями не позволяет восстановить исходное сообщение полностью, но дает достаточное на практике приближение к нему.
